# Task 1 — Fetch and Validate the Response

```python
import requests
import json
import pandas as pd
```

```python

url = "https://restcountries.com/v3.1/all?fields=name,population,region,area"
try:

  responce = requests.get(url, timeout= 10)
  if responce.status_code == 200:
    data = responce.json()
    print("Total Number of countries retrieved: ", len(data))
except:
  print("Descriptive Error Message")
```

#Task 2 — Extract and Display Key Fields

```python
responce = requests.get(url, timeout= 10)
data = responce.json()
```

```python

df = pd.DataFrame({
            "country_name": [c["name"]["common"] for c in data],
            "population": [c.get("population") for c in data],
            "region": [c.get("region") for c in data],
            "area": [c.get("area") for c in data],
        })
df.head(10)
```

# Task 3 — Summarize the Data

```python
high_population = df.groupby("region")["population"].sum().idxmax()
print("The highest population is: ",high_population)
```

```python
largest_area =  df.groupby("country_name")["area"].sum().idxmax()
print("The Largest Area is: ",largest_area)
```
